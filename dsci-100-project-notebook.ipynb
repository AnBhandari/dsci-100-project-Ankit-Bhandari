```python
# The data is already tidy, since each row contains one team in one year, 
# each column is a single variable, and each cell has one value.
# No melt or pivot methods are needed

# Creation of categorical target, which is the question made in task 2.1
# Remove the 8th seed placeholder teams.
march_madness = data.loc[data["seed"] != 0]
march_madness = march_madness.assign(
    final_four = march_madness["tourney_finish"].isin(["top_4_loss", "top_2_loss", "champ"])
)
march_madness["final_four"] = march_madness["final_four"].replace({True: "yes", False: "no"})

# Selecting for the pre tounament tournament predictors and the categorical target.
# Avoid including columns containing the tournament results, since it would reveal the answer.
march_madness = march_madness[["seed", "reg_wins_pct", "final_four"]]
march_madness.head(6)
```

```python
import altair as alt

reg_pct_hist = alt.Chart(
    march_madness,
    title="Distribution of Regular-Season Win Percentage"
).mark_bar().encode(
    x = alt.X("reg_wins_pct").bin(maxbins=30).title("Regular-Season Win Percentage (%)"),
    y = alt.Y("count()").title("Number of Teams"),
)
reg_pct_hist 
```

```python
ff_props = march_madness["final_four"].value_counts(normalize=True).reset_index()

ff_bar = alt.Chart(
    ff_props,
    title="Proportion of Teams Reaching the Final Four"
).mark_bar().encode(
    x = alt.X("final_four").title("Has Reached the Final Four?"),
    y = alt.Y("proportion").title("Proportion of Teams")
).properties(width=200)
ff_bar
```

The plot shows how seed and regular-season win percentages are related to each other and to being apart of the Final Four. On the plot, Final Four teams are clustered in the top-left. These are the low seeds with high win percentages, information which can be utilized for a classifier.

```python
scatter = alt.Chart(
    march_madness,
    title="Seed vs Regular-Season Win Percentage"
).mark_circle(opacity=0.4).encode(
    x = alt.X("seed").title("Tournament Seed"),
    y = alt.Y("reg_wins_pct").scale(zero=False).title("Regular-Season Win Percentage (%)"),
    color = alt.Color("final_four").title("Reached Final Four")
).properties(width=300)
scatter
```

The plot shows what teams at each seed reached the Final Four. About 56% of first seeds made it, and any teams seeded at 10th or lower never made it to the Final Four. This demonstrates that seed is a strong predictor, and highlights the rarity of "yes" relative to all the teams there are.

```python
final_four_seed = (
    march_madness
    .assign(reached=march_madness["final_four"] == "yes")
    .groupby("seed")
    .agg(prop_final_four=("reached", "mean"))
    .reset_index()
)
final_four_chart = alt.Chart(
    final_four_seed,
    title="Proportion of Teams Which Have Reached the Final Four"
).mark_bar().encode(
    x = alt.X("seed:O").title("Tournament Seed"),
    y = alt.Y("prop_final_four").title("Proportion Reaching the Final Four")
)
final_four_chart
```
